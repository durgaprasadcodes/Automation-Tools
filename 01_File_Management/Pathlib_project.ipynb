# Creating a Simple Automation Tool For Files 

```python
from pathlib import Path
import os
import getpass
```

```python
class FileSystemTool:
    
    def __init__(self):
        self.root = Path.cwd()
        
    def show_current_directory(self):
        return f"Current Directory {self.root}"
    
    def list_files(self):
        for item in self.root.iterdir():
            if item.is_file():
                print(f"📄 {item.name}")

            elif item.is_dir():
                print(f"📁 {item.name}")
                
    def find_files(self,extension):
        files = self.root.rglob(f"*.{extension}")
        for file in files:
            print("File Name : ",file.name)
            print("File Size : ",file.stat().st_size)
            
    def create_folder(self,name):
        folder = self.root/name
        folder.mkdir(exist_ok=True)
        print(f"Folder Created: {folder}")
        
    def create_file(self,name,content):
        file = self.root/name
        file.parent.mkdir(exist_ok=True,parent=True)
        file.write_text(content)
        print("File Created",file.name)
        
    def read_file(self, name):
        file = self.root / name
        if file.exists():
            return file.read_text()
        return "File does not exist."

    def delete_file(self, name):
        file = self.root / name
        if file.exists() and file.is_file():
            file.unlink()
            return f"{file.name} - File Deleted Successfully"
        return f"Cannot delete: {name} is not an existing file."
    
    def get_environment_variable(self, name):
        return os.getenv(name)

    def current_system_user(self):
        return getpass.getuser()
```

```python
agent_fs = FileSystemTool()

agent_fs.show_current_directory()

agent_fs.list_files()

agent_fs.current_system_user()

agent_fs.delete_file("test.py")
```
